```python
import pandas as pd
df = pd.read_csv('Financial_Data_2023_2025.csv')
df.columns = df.columns.str.strip()
df = df.sort_values(by=['Company', 'Year'])
df['Revenue Growth (%)'] = df.groupby('Company')['Total Revenue'].pct_change() * 100
df['Net Income Growth (%)'] = df.groupby('Company')['Net Income'].pct_change() * 100
df['Net Profit Margin (%)'] = (df['Net Income'] / df['Total Revenue']) * 100
print(df[['Company', 'Year', 'Total Revenue', 'Revenue Growth (%)', 'Net Profit Margin (%)']])
```

# 📊Executive summary & financial analysis(2023 - 2025):

## 1. Methodology & data sources
Financial metrics were extracted from official **SEC 10-K filings** for **Microsoft (MSFT)**, **Apple (AAPL)**, and **Tesla (TSLA)** covering the fiscal years 2023 to 2025. Key evaluated metrics include Total Revenue, Net Income, Total Assets, Total Liabilities, and Cash Flow from Operating Activities.

---

## 2. Key Insights & Financial Trends

* **Microsoft (MSFT): Exceptional Growth & Stability**
  * **Revenue & Income:** Demonstrated strong, consistent double-digit growth driven by Enterprise Cloud and AI integration.
  * **Cash Flow:** Operating Cash Flow steadily increased, showing strong cash conversion and financial health.

* **Apple (AAPL): Massive Scale & High Cash Generation**
  * **Profitability:** Maintained massive net income and robust operating cash flows (~$110B–$125B annually).
  * **Margins:** Maintained extremely strong profit margins with controlled liability growth.

* **Tesla (TSLA): Margin Volatility & Revenue Expansion**
  * **Revenue:** Continued top-line revenue growth year-over-year.
  * **Profitability:** Net income experienced margin compression in 2024 due to EV price reductions and strategic reinvestments, before showing recovery signs.

---

## 3. Implications for AI Financial Chatbot Development
* **Query Optimization:** The chatbot must be trained to differentiate between top-line growth (Revenue) and bottom-line efficiency (Net Income margin).
* **Multi-Year Context:** Enabling the AI to compute YoY (Year-over-Year) growth rate percentages dynamically using `pct_change()` provides richer answers than reporting static numbers.
* **Risk Assessment:** The model should highlight trends in Cash Flow from Operations as a primary indicator of liquidity and financial sustainability.
